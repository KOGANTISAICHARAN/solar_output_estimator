```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.compose import TransformedTargetRegressor
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
pd.set_option("display.width", 200, "display.max_columns", 30)
```

```python
TARGET = "AC_POWER_PLANT"
```

```python
df = pd.read_csv("solar_daytime_training_dataset.csv")
print("Rows, columns:", df.shape)
print(df.head())
print(df.dtypes)
print(df["PLANT_ID"].value_counts())
```

```python
print("Missing values:", df.isna().sum().sum())
print("Duplicate rows (same plant and time):", df.duplicated(["PLANT_ID", "DATE_TIME"]).sum())
```

```python
df["DATE_TIME"] = pd.to_datetime(df["DATE_TIME"])
df = df.sort_values(["PLANT_ID", "DATE_TIME"]).reset_index(drop=True)
df["DATE"] = df["DATE_TIME"].dt.date
print("Period:", df["DATE_TIME"].min(), "to", df["DATE_TIME"].max(), "|", df["DATE"].nunique(), "days")
```

```python
same_value = [c for c in df.columns if df[c].nunique() == 1]
print("Dropped useless columns:", same_value)
df = df.drop(columns=same_value)
```

```python
print("Negative power rows:", (df[TARGET] < 0).sum())
print("Negative irradiation rows:", (df["IRRADIATION"] < 0).sum())
print("Check TEMP_GAP = module - ambient:",
      np.allclose(df["TEMP_GAP"], df["MODULE_TEMPERATURE"] - df["AMBIENT_TEMPERATURE"]))
print("Check IRRAD_X_MODTEMP = irradiation * module temp:",
      np.allclose(df["IRRAD_X_MODTEMP"], df["IRRADIATION"] * df["MODULE_TEMPERATURE"]))
hour = df["DATE_TIME"].dt.hour + df["DATE_TIME"].dt.minute / 60
print("Check HOUR_SIN matches the time:", np.allclose(df["HOUR_SIN"], np.sin(2 * np.pi * hour / 24)))
print("Check PLANT_ENC matches PLANT_ID:",
      (df.groupby("PLANT_ID")["PLANT_ENC"].nunique() == 1).all())
```

```python
step = df.groupby(["PLANT_ID", "DATE"])["DATE_TIME"].diff()
print("Missing 15-min steps within days:", int((step.dropna() > pd.Timedelta("15min")).sum()))
```

```python
df["POWER_PER_SUN"] = df[TARGET] / df["IRRADIATION"]
low = df.groupby("PLANT_ID")["POWER_PER_SUN"].transform(lambda s: s.quantile(0.01))
df["UNUSUAL_LOW"] = (df["POWER_PER_SUN"] < low) & (df["IRRADIATION"] > 0.1)
print("Flagged rows (very low power for the sunlight, kept in the data):", int(df["UNUSUAL_LOW"].sum()))
```

```python
days = sorted(df["DATE"].unique())
n = len(days)
train_days = days[:int(n * 0.70)]
val_days = days[int(n * 0.70):int(n * 0.85)]
test_days = days[int(n * 0.85):]
train = df[df["DATE"].isin(train_days)]
val = df[df["DATE"].isin(val_days)]
test = df[df["DATE"].isin(test_days)]
print("Train:", len(train), "rows,", len(train_days), "days |",
      "Validation:", len(val), "rows,", len(val_days), "days |",
      "Test:", len(test), "rows,", len(test_days), "days")
```

```python
print(df[[TARGET, "IRRADIATION", "MODULE_TEMPERATURE", "AMBIENT_TEMPERATURE"]].describe().round(2))
print(df.groupby("PLANT_ID")[TARGET].describe().round(1))
```

```python
plt.figure(figsize=(6, 4))
for plant, g in df.groupby("PLANT_ID"):
    plt.scatter(g["IRRADIATION"], g[TARGET], s=5, alpha=0.4, label=plant)
plt.xlabel("Irradiation"); plt.ylabel("AC power (kW)")
plt.title("Power vs irradiation"); plt.legend(); plt.show()

plt.figure(figsize=(6, 4))
df.groupby(df["DATE_TIME"].dt.hour)[TARGET].mean().plot()
plt.xlabel("Hour of day"); plt.ylabel("Mean power (kW)")
plt.title("Average power by hour"); plt.show()

cols = [TARGET, "IRRADIATION", "MODULE_TEMPERATURE", "AMBIENT_TEMPERATURE", "HOUR_SIN", "HOUR_COS"]
print(df[cols].corr().round(2))
```

```python
gap = df.groupby(["PLANT_ID", "DATE"])["DATE_TIME"].diff()
change = df.groupby(["PLANT_ID", "DATE"])["IRRADIATION"].diff()
df["IRR_CHANGE"] = change.where(gap == pd.Timedelta("15min"), 0.0)

FEATURES = ["IRRADIATION", "MODULE_TEMPERATURE", "AMBIENT_TEMPERATURE", "TEMP_GAP",
            "IRRAD_X_MODTEMP", "HOUR_SIN", "HOUR_COS", "PLANT_ENC", "IRR_CHANGE"]
```

```python
train = df[df["DATE"].isin(train_days)]
val = df[df["DATE"].isin(val_days)]
test = df[df["DATE"].isin(test_days)]
X_train, y_train = train[FEATURES], train[TARGET]
X_val, y_val = val[FEATURES], val[TARGET]
X_test, y_test = test[FEATURES], test[TARGET]
```

```python
models = {
    "Linear Regression": make_pipeline(StandardScaler(), LinearRegression()),
    "Random Forest": RandomForestRegressor(n_estimators=200, min_samples_leaf=3, random_state=42),
    "SVR": TransformedTargetRegressor(
        regressor=make_pipeline(StandardScaler(), SVR(C=10)),
        transformer=StandardScaler()),
}
```

```python
def evaluate(model, X, y):
    pred = model.predict(X)
    return {"MAE": mean_absolute_error(y, pred),
            "RMSE": np.sqrt(mean_squared_error(y, pred)),
            "R2": r2_score(y, pred)}


results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    results[name] = evaluate(model, X_val, y_val)
results = pd.DataFrame(results).T.round(3)
print(results)

best = results["RMSE"].idxmin()
print("Selected model:", best)
```

```python
final_model = models[best]
final_model.fit(X_train, y_train)
print("Trained:", best, "on", len(X_train), "rows")
```

```python
table = pd.DataFrame({
    "validation": evaluate(final_model, X_val, y_val),
    "test": evaluate(final_model, X_test, y_test),
}).T.round(3)
print(table)

for plant, g in test.groupby("PLANT_ID"):
    m = evaluate(final_model, g[FEATURES], g[TARGET])
    print("Test, plant", plant, {k: round(float(v), 3) for k, v in m.items()})
```

```python
pred_test = final_model.predict(X_test)
fig, axs = plt.subplots(1, 2, figsize=(11, 4))
axs[0].scatter(y_test, pred_test, s=5, alpha=0.4)
axs[0].plot([0, y_test.max()], [0, y_test.max()], "r--")
axs[0].set(xlabel="Actual kW", ylabel="Predicted kW", title="Test days: actual vs predicted")
axs[1].hist(y_test - pred_test, bins=50)
axs[1].set(xlabel="Actual - predicted (kW)", title="Prediction errors (residuals)")
plt.tight_layout(); plt.show()
```

```python
param_grids = {
    "SVR": {"regressor__svr__C": [1, 3, 10, 30, 100],
            "regressor__svr__epsilon": [0.01, 0.02, 0.05, 0.1],
            "regressor__svr__gamma": ["scale", 0.02, 0.05]},
    "Random Forest": {"n_estimators": [200, 400],
                      "max_depth": [None, 10, 20],
                      "min_samples_leaf": [1, 3, 5]},
}
```

```python
if best not in param_grids:
    print(best, "has no settings worth tuning. Keep the model from Step 7.")
    tuned_model = final_model
else:
    train_sorted = train.sort_values("DATE_TIME")        
    search = GridSearchCV(models[best], param_grids[best],
                          cv=TimeSeriesSplit(n_splits=3),
                          scoring="neg_root_mean_squared_error", n_jobs=-1)
    search.fit(train_sorted[FEATURES], train_sorted[TARGET])
    print("Best settings:", search.best_params_)
    print("Best CV RMSE:", round(-search.best_score_, 1))
    tuned_model = search.best_estimator_
```

```python
before = evaluate(final_model, X_val, y_val)["RMSE"]
after = evaluate(tuned_model, X_val, y_val)["RMSE"]
print(f"Validation RMSE  before tuning: {before:.1f}   after tuning: {after:.1f}")
if after < before:
    final_model = tuned_model
    print("Using the tuned model.")
else:
    print("Tuning did not help on validation, keeping the Step 7 model.")

print("\nFINAL RESULT on unseen test days")
print({k: round(float(v), 3) for k, v in evaluate(final_model, X_test, y_test).items()})
```

```python
from sklearn.base import clone
import numpy as np
import pandas as pd

# Sort the training data
train_time = train_time.sort_values("DATE_TIME")

# Split training period into model-training and calibration data
unique_train_dates = sorted(train_time["DATE_TIME"].unique())

calibration_cutoff = unique_train_dates[
    int(len(unique_train_dates) * 0.80)
]

model_training = train_time[
    train_time["DATE_TIME"] <= calibration_cutoff
].copy()

calibration_data = train_time[
    train_time["DATE_TIME"] > calibration_cutoff
].copy()

# Prepare training inputs
X_model_training = model_training[FEATURES]
y_model_training = model_training[TARGET]

# Prepare calibration inputs
X_calibration = calibration_data[FEATURES]
y_calibration = calibration_data[TARGET]

# Train a fresh copy of our selected model
calibration_model = clone(final_model)

calibration_model.fit(
    X_model_training,
    y_model_training
)

# Predict calibration readings
calibration_predictions = calibration_model.predict(
    X_calibration
)

# Create calibration results
calibration_results = pd.DataFrame({
    "PLANT_ENC": calibration_data["PLANT_ENC"].values,
    "ACTUAL": y_calibration.values,
    "PREDICTED": calibration_predictions
})

# Calculate prediction errors
calibration_results["ABS_ERROR"] = (
    calibration_results["ACTUAL"] -
    calibration_results["PREDICTED"]
).abs()

# Calculate separate 95th percentile thresholds for each plant
anomaly_thresholds = (
    calibration_results
    .groupby("PLANT_ENC")["ABS_ERROR"]
    .quantile(0.95)
)

print("Anomaly thresholds:")
print(anomaly_thresholds)

print("\nCalibration data shape:")
print(calibration_data.shape)
```

```python
# Copy the chronological test data
anomaly_results = test_time[
    ["DATE_TIME", "PLANT_ID", "PLANT_ENC",
     "IRRADIATION", "AC_POWER_PLANT"]
].copy()

# Add model predictions
anomaly_results["PREDICTED_AC_POWER"] = time_predictions

# Calculate signed prediction error
anomaly_results["ERROR"] = (
    anomaly_results["AC_POWER_PLANT"] -
    anomaly_results["PREDICTED_AC_POWER"]
)

# Calculate absolute error
anomaly_results["ABS_ERROR"] = (
    anomaly_results["ERROR"].abs()
)

# Add plant-specific anomaly thresholds
anomaly_results["THRESHOLD"] = (
    anomaly_results["PLANT_ENC"].map(
        anomaly_thresholds.to_dict()
    )
)

# Identify anomalies
anomaly_results["ANOMALY_STATUS"] = np.where(
    anomaly_results["ABS_ERROR"] >
    anomaly_results["THRESHOLD"],
    "ANOMALY",
    "NORMAL"
)

# Identify deviation direction
anomaly_results["DEVIATION_TYPE"] = np.where(
    anomaly_results["ERROR"] < 0,
    "UNDERPERFORMANCE",
    "ABOVE_EXPECTED"
)

# Display results
print(
    anomaly_results[
        [
            "DATE_TIME",
            "PLANT_ID",
            "AC_POWER_PLANT",
            "PREDICTED_AC_POWER",
            "ABS_ERROR",
            "THRESHOLD",
            "ANOMALY_STATUS",
            "DEVIATION_TYPE"
        ]
    ].head(20).to_string(index=False)
)
```
